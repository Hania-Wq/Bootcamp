## 2. Quick-start with Pipelines

The `pipeline` API wraps tokenization, model loading, and inference in one object.

### 2.1 Sentiment Analysis

```python
from transformers import pipeline

# 1. Load a sentiment-analysis pipeline (defaults to a small model)
sentiment = pipeline("sentiment-analysis")

# 2. Run inference
examples = [
    "Hugging Face makes NLP super accessible!",
    "I dislike bugs in my code..."
]
results = sentiment(examples)
for text, res in zip(examples, results):
    print(f"{text!r:50} → label={res['label']}, score={res['score']:.3f}")
```

### 2.2 Text Generation

```python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")
prompt = "Once upon a time"
out = generator(prompt, max_length=30, num_return_sequences=1)
print(out[0]["generated_text"])
```

### 2.3 Question Answering

```python
from transformers import pipeline

qa = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")
context = (
    "Hugging Face is an AI company with the mission to democratize good machine learning. "
    "The Transformers library provides thousands of pretrained models in 100+ languages."
)

answer = qa({
    "question": "What is the mission of Hugging Face?",
    "context": context
})
print(answer["answer"])
```

# 3. Assignment: Build Your Own Pipeline

**1. Choose one problem from the list below.**

Text summarization of news articles

**2. Find a suitable pre-trained model on Hugging Face Models.**

facebook/bart-large-cnn (choice made based on summarization sorted with trending / most downloads on hugging face)

````python
#3. Create a pipeline in Colab to solve the problem—and demonstrate it on 2–3 examples.

from transformers import pipeline

summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

articles = {
    "tech": (
        "Apple unveiled its latest iPhone lineup on Tuesday, featuring a new A19 chip "
        "that the company claims is 20% faster than its predecessor. The phones also "
        "include an upgraded camera system with improved low-light performance and a "
        "redesigned titanium frame. Analysts say the launch comes at a critical time "
        "for Apple, as smartphone sales have slowed globally and competition from "
        "Chinese manufacturers has intensified. The new models will be available in "
        "stores starting next Friday, with prices starting at $999."
    ),
    "climate": (
        "A new report released by the United Nations warns that global temperatures "
        "could rise by more than 2.5 degrees Celsius by the end of the century if "
        "current emissions trends continue. The report, compiled by hundreds of "
        "scientists from over 60 countries, calls for immediate and drastic cuts to "
        "fossil fuel use. It highlights that developing nations are likely to bear "
        "the brunt of climate-related disasters despite contributing the least to "
        "global emissions. World leaders are expected to discuss the findings at the "
        "upcoming climate summit next month."
    ),
    "sports": (
        "In a thrilling final match, the underdog team clinched the championship title "
        "after a dramatic penalty shootout, ending a 15-year title drought for the club. "
        "The match had remained scoreless through regulation and extra time, with both "
        "goalkeepers making crucial saves throughout the game. Thousands of fans flooded "
        "the streets of the city in celebration following the victory, which many are "
        "already calling one of the greatest finals in the tournament's history."
    ),
}

for name, text in articles.items():
    summary_text = summarizer(text, max_length=60, min_length=20, do_sample=False)[0]["summary_text"]

    original_words = len(text.split())
    summary_words = len(summary_text.split())
    original_chars = len(text)
    summary_chars = len(summary_text)
    compression_ratio = round(summary_words / original_words * 100, 1)

    print(f"--- {name.upper()} ---")
    print("Summary:", summary_text)
    print(f"Original: {original_words} words / {original_chars} chars")
    print(f"Summary:  {summary_words} words / {summary_chars} chars")
    print(f"Compression: {compression_ratio}% of original length")
    print()
````

````python
#4. Experiment with model parameters (e.g. max_length, top_k, temperature) and compare results.

text = articles["tech"]

for max_len, min_len in [(15, 5), (60, 20), (120, 60)]:
    summary = summarizer(text, max_length=max_len, min_length=min_len, do_sample=False)
    print(f"max_length={max_len}, min_length={min_len}:")
    print(summary[0]["summary_text"])
    print()
````

````python
for beams in [1, 4, 8]:
    summary = summarizer(text, max_length=60, min_length=20, num_beams=beams, do_sample=False)
    print(f"num_beams={beams}:")
    print(summary[0]["summary_text"])
    print()
````

````python
for temp in [0.3, 0.7, 1.2]:
    summary = summarizer(text, max_length=60, min_length=20, do_sample=True, temperature=temp)
    print(f"temperature={temp}:")
    print(summary[0]["summary_text"])
    print()
````

# Machine translation

````python
# Machine Translation: English → Mandarin (with Pinyin)

!pip install pypinyin -q

from transformers import pipeline
from pypinyin import pinyin, Style

translator = pipeline("translation", model="Helsinki-NLP/opus-mt-en-zh")

examples = [
    "Hello, my name is Li Ming.",
    "I am a student.",
    "Today is Monday.",
    "I like drinking tea.",
    "This is my friend.",
    "I have two cats.",
]

print("ENGLISH → MANDARIN TRANSLATION (with Pinyin)")
print()

for text in examples:
    chinese = translator(text)[0]["translation_text"]
    pinyin_text = " ".join([s[0] for s in pinyin(chinese, style=Style.TONE)])

    print(f"EN: {text}")
    print(f"ZH: {chinese}")
    print(f"Pinyin: {pinyin_text}")
    print(f"(EN length: {len(text.split())} words | ZH length: {len(chinese)} chars)")
    print("-" * 60)

````

````python
sample_text = "I have two cats and a dog, and they like to play with a ball every day."

# --- num_beams comparison ---
print("\n--- Effect of num_beams ---")
for beams in [1, 4, 8]:
    result = translator(sample_text, num_beams=beams)[0]["translation_text"]
    py = " ".join([s[0] for s in pinyin(result, style=Style.TONE)])
    print(f"num_beams={beams}: {result}  ({py})")

# --- max_length comparison ---
print("\n--- Effect of max_length ---")
for max_len in [10, 30, 60]:
    result = translator(sample_text, max_length=max_len)[0]["translation_text"]
    py = " ".join([s[0] for s in pinyin(result, style=Style.TONE)])
    print(f"max_length={max_len}: {result}  ({py})")

# --- do_sample + temperature comparison ---
print("\n--- Effect of do_sample + temperature ---")
for temp in [0.3, 0.7, 1.2]:
    result = translator(sample_text, do_sample=True, temperature=temp)[0]["translation_text"]
    py = " ".join([s[0] for s in pinyin(result, style=Style.TONE)])
    print(f"temperature={temp}: {result}  ({py})")
````
