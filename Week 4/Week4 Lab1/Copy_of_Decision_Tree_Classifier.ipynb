#Decision Tree Classifier Lab


#Load essential libraries

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn import tree
from sklearn.model_selection import train_test_split
```

#Import the dataset

```python
df = pd.read_csv("/titanic.csv")
df
```

##Drop meaningless information

```python
df = df.drop(columns = ['PassengerId','Name','SibSp', 'Parch', 'Ticket', 'Cabin', 'Embarked'], axis = 'columns')
df
```

##Data Preprocessing

```python
import math
mean_age = df.Age.mean()
mean_age = math.floor(mean_age)
print(mean_age)
```

```python
df['Age'] = df.Age.fillna(mean_age)
df
```

##Label Encode the non numeric features

```python
LE_sex = LabelEncoder()
df['Sex'] = LE_sex.fit_transform(df['Sex'])

df
```

## Define Feature Matrix (X) and Target Vector (y)


```python
y = df['Survived']
X = df.drop('Survived', axis = 'columns')
X
```

## Train-Test Split and Model Initialization


```python
model = tree.DecisionTreeClassifier()

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.4)
```

##Train the model

```python
model.fit(X_train, y_train)

model.score(X_test, y_test)
```

##Predict results

```python
model.predict(X_test)
```

```python
y_test
```

```python
#load dataset
df = pd.read_csv("/titanic.csv")
df.head()
```

```python
df.isna().sum()
```

```python
#dropping unnecessary columns
df = df.drop(columns = ['PassengerId','Ticket', 'Cabin',], axis = 'columns')
df
```

```python
#filling NaN
mean_age = df.Age.mean()
mean_age = math.floor(mean_age)
df['Age'] = df.Age.fillna(mean_age)
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
print(df.isna().sum())
```

```python
#Feature engineering Title
df['Title'] = df['Name'].str.extract(r',\s*([^.]*)\.')
df['Title'] = df['Title'].replace(['Lady', 'Countess','Capt', 'Col',
    'Don', 'Dr', 'Major', 'Rev', 'Sir', 'Jonkheer', 'Dona'], 'Rare')
df['Title'] = df['Title'].replace('Mlle', 'Miss')
df['Title'] = df['Title'].replace('Ms', 'Miss')
df['Title'] = df['Title'].replace('Mme', 'Mrs')

#Feature engineering FamilySize and IsAlone
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df['IsAlone'] = (df['FamilySize'] == 1).astype(int)
```

```python
df.drop(['Name'], axis=1, inplace=True)
df.head()
```

```python
#Encoding Features
LE_sex = LabelEncoder()
df['Sex'] = LE_sex.fit_transform(df['Sex'])

ohe = OneHotEncoder(drop='first', sparse_output=False)
encoded = ohe.fit_transform(df[['Embarked', 'Title']])

encoded_df = pd.DataFrame(encoded, columns=ohe.get_feature_names_out(['Embarked', 'Title']))

df = df.drop(['Embarked', 'Title'], axis=1).reset_index(drop=True)
df = pd.concat([df, encoded_df], axis=1)

df.head()
```

```python
y = df['Survived']
X = df.drop('Survived', axis = 'columns')
```

```python
model = tree.DecisionTreeClassifier()
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25)

model.fit(X_train, y_train)
model.score(X_test, y_test)
```

```python
model.predict(X_test)
```
