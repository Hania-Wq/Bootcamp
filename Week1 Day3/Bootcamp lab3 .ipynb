## step 4 data analysis

```python
#Calculate daily average temperature
avg_temp= df["temperature_C"].mean()
print("average temperature:",avg_temp)

#Find max, min, mean for each metric
print("\n min values:\n" , df.min())
print("\n max values:\n", df.max())
print("\n mean values:\n", df.select_dtypes(include='number').mean())
```

```python
# Which hour of the day is the most humid on average?
df["timestamp"] = pd.to_datetime(df["timestamp"])   
df["hour"] = df["timestamp"].dt.hour                
hourly_avg_humidity = df.groupby("hour")["humidity_%"].mean()
print(hourly_avg_humidity)
most_humid_hour = hourly_avg_humidity.idxmax()
print("Most humid hour on average:", most_humid_hour, "\nhumidity%:", hourly_avg_humidity.max())
```

## Numpy Matrix Exercises

```python
# TODO: Extract temperature and wind_speed as NumPy arrays
import numpy as np

temp = df['temperature_C'].to_numpy()
print("temp:", temp)
wind = df['wind_speed_kmph'].to_numpy()
print("wind:", wind)
```

```python
#a) Reshape into matrix form
#Assume each row is a day
#Reshape temperature into a (10, 24) matrix
temp_matrix = temp.reshape((10, 24))
print(temp_matrix, "\n")
#Calculate daily min, max, and mean using axis-based operations
dailymin=temp_matrix.min(axis=1)
print("dailymin:", dailymin)
dailymax=temp_matrix.max(axis=1)
print("dailymax:", dailymax)
dailymean=temp_matrix.mean(axis=1)
print("dailymean:", dailymean)
```

```python
# b) Normalize the temperature matrix
#Subtract the mean and divide by std deviation
#Do it manually using NumPy functions
def normalize(matrix):
    normalized_matrix=(matrix-matrix.mean())/matrix.std()
    return normalized_matrix
normalized_temp=normalize(temp_matrix)
print(normalized_temp)
```

```python
#c) Apply custom mask/filter
# TODO: Create boolean mask and filter wind speeds
mask = wind > 15
high_wind = wind[mask]

print("mask:", mask)
print("\nhigh wind readings:", high_wind)
print("\nnumber of high wind readings:", high_wind.size)
```

## Final Challenge: Write Your Own Function

```python
#Write a function `daily_summary(matrix)`
humidity_matrix = df['humidity_%'].to_numpy().reshape((10, 24))
pressure_matrix = df['pressure_hPa'].to_numpy().reshape((10, 24))
visibility_matrix = df['visibility_km'].to_numpy().reshape((10, 24))
wind_matrix = wind.reshape((10, 24))
```

```python
def daily_summary(temp_matrix, humidity_matrix, wind_matrix, pressure_matrix, visibility_matrix):
    summaries = []
    num_days = temp_matrix.shape[0]  # 10

    for i in range(num_days):
        summary = {
            "temperature": {
                "min": temp_matrix[i].min(),
                "max": temp_matrix[i].max(),
                "mean": temp_matrix[i].mean()
            },
            "humidity": {
                "min": humidity_matrix[i].min(),
                "max": humidity_matrix[i].max(),
                "mean": humidity_matrix[i].mean()
            },
            "wind_speed": {
                "min": wind_matrix[i].min(),
                "max": wind_matrix[i].max(),
                "mean": wind_matrix[i].mean()
            },
            "pressure": {
                "min": pressure_matrix[i].min(),
                "max": pressure_matrix[i].max(),
                "mean": pressure_matrix[i].mean()
            },
            "visibility": {
                "min": visibility_matrix[i].min(),
                "max": visibility_matrix[i].max(),
                "mean": visibility_matrix[i].mean()
            }
        }
        summaries.append(summary)

    return summaries

summaries = daily_summary(temp_matrix, humidity_matrix, wind_matrix, pressure_matrix, visibility_matrix)
for i, s in enumerate(summaries):
    print(f"Day {i+1}: {s}\n")
```
