## Step 1: Download and clean the book text

```python
!wget -q https://www.gutenberg.org/files/2554/2554-0.txt -O crime_and_punishment.txt
!ls -la crime_and_punishment.txt
```

```python
with open("crime_and_punishment.txt", "r", encoding="utf-8") as f:
    raw_text = f.read()

start_marker = "*** START OF THE PROJECT GUTENBERG EBOOK"
end_marker = "*** END OF THE PROJECT GUTENBERG EBOOK"

start_idx = raw_text.find(start_marker)
end_idx = raw_text.find(end_marker)

book_text = raw_text[start_idx:end_idx]

print(f"Total characters: {len(book_text)}")
print(book_text[:500])
```

## Step 2: Chunk the text

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150,
    separators=["\n\n", "\n", ". ", " ", ""]
)

chunks = splitter.split_text(book_text)

print(f"Number of chunks: {len(chunks)}")
print(chunks[5])
```

## Step 3: Embed chunks and build the vector store

I used a free local embedding model (`sentence-transformers`) instead of a paid API,
so all Gemini token quota is reserved for the LLM generation step later.

```python
import chromadb
from sentence_transformers import SentenceTransformer

embedder = SentenceTransformer("all-MiniLM-L6-v2")

client = chromadb.PersistentClient(path="./chroma_db")

# Recreate the collection fresh each run of this cell
try:
    client.delete_collection(name="crime_and_punishment")
except Exception:
    pass

collection = client.create_collection(name="crime_and_punishment")

batch_size = 64
for i in range(0, len(chunks), batch_size):
    batch = chunks[i:i+batch_size]
    embeddings = embedder.encode(batch).tolist()
    ids = [f"chunk_{i+j}" for j in range(len(batch))]
    collection.add(documents=batch, embeddings=embeddings, ids=ids)
    print(f"Embedded {min(i + batch_size, len(chunks))} / {len(chunks)}")

print("Done! Local vector store built and saved to disk (./chroma_db).")
```

## Step 4: Retrieval and generation functions


```python
def retrieve_context_chunks(query: str, top_k: int = 5) -> list[str]:
    query_embedding = embedder.encode([query]).tolist()
    results = collection.query(query_embeddings=query_embedding, n_results=top_k)
    return results["documents"][0]  # list of separate chunk strings


def generate_answer(query: str, chat_history=None) -> str:
    context_chunks = retrieve_context_chunks(query, top_k=5)
    context = "\n\n---\n\n".join(context_chunks)

    history_text = ""
    if chat_history:
        for turn in chat_history:
            role = "User" if turn["role"] == "user" else "Assistant"
            history_text += f"{role}: {turn['content']}\n"

    prompt = f"""You are answering questions about the novel "Crime and Punishment" by Fyodor Dostoevsky.
Use ONLY the context below to answer. If the answer isn't in the context, say you don't know.

Context:
{context}

Conversation so far:
{history_text}
User: {query}
Assistant:"""

    # Using the lite model here to preserve free-tier quota on the main chat model
    gen_model = genai.GenerativeModel(model_name="models/gemini-flash-lite-latest")
    response = gen_model.generate_content(prompt)
    return response.text


# quick sanity test
sample_query = "Why did Raskolnikov commit the murder?"
print(generate_answer(sample_query))
```

## Step 5: Streamlit app — assignment Steps 1 & 2

Single-turn and multi-turn (in-memory) modes, toggled from the sidebar.

```python
%%writefile app.py
import streamlit as st
import chromadb
from sentence_transformers import SentenceTransformer
import google.generativeai as genai
import os

st.set_page_config(page_title="Crime and Punishment RAG", page_icon="📖")
st.title("📖 Crime and Punishment — RAG Chat")

@st.cache_resource
def load_resources():
    embedder = SentenceTransformer("all-MiniLM-L6-v2")
    client = chromadb.PersistentClient(path="./chroma_db")
    collection = client.get_collection(name="crime_and_punishment")
    genai.configure(api_key=os.environ["GEMINI_API_KEY"])
    gen_model = genai.GenerativeModel(model_name="models/gemini-flash-lite-latest")
    return embedder, collection, gen_model

embedder, collection, gen_model = load_resources()

def retrieve_context(query, top_k=5):
    query_embedding = embedder.encode([query]).tolist()
    results = collection.query(query_embeddings=query_embedding, n_results=top_k)
    return "\n\n---\n\n".join(results["documents"][0])

def generate_answer(query, chat_history=None):
    context = retrieve_context(query)
    history_text = ""
    if chat_history:
        for turn in chat_history:
            role = "User" if turn["role"] == "user" else "Assistant"
            history_text += f"{role}: {turn['content']}\n"

    prompt = f"""You are answering questions about the novel "Crime and Punishment" by Fyodor Dostoevsky.
Use ONLY the context below to answer. If the answer isn't in the context, say you don't know.

Context:
{context}

Conversation so far:
{history_text}
User: {query}
Assistant:"""
    response = gen_model.generate_content(prompt)
    return response.text

mode = st.sidebar.radio("Mode", ["Step 1: Single-turn", "Step 2: Multi-turn (in-memory)"])

if "messages" not in st.session_state:
    st.session_state.messages = []

for msg in st.session_state.messages:
    with st.chat_message(msg["role"]):
        st.markdown(msg["content"])

user_input = st.chat_input("Ask something about Crime and Punishment...")

if user_input:
    with st.chat_message("user"):
        st.markdown(user_input)

    history = st.session_state.messages if mode.startswith("Step 2") else None

    with st.spinner("Thinking..."):
        answer = generate_answer(user_input, chat_history=history)

    with st.chat_message("assistant"):
        st.markdown(answer)

    st.session_state.messages.append({"role": "user", "content": user_input})
    st.session_state.messages.append({"role": "assistant", "content": answer})
```

## Step 6: Streamlit app — assignment Step 3 (persistent history via Supabase)

Kept as a **separate file** (`app_step3.py`) from `app.py` so both stages of the
build remain visible and comparable.


```python
%%writefile app_step3.py
import streamlit as st
import chromadb
from sentence_transformers import SentenceTransformer
import google.generativeai as genai
from supabase import create_client
import os

st.set_page_config(page_title="Crime and Punishment RAG", page_icon="📖")
st.title("📖 Crime and Punishment — RAG Chat")

@st.cache_resource
def load_resources():
    embedder = SentenceTransformer("all-MiniLM-L6-v2")
    client = chromadb.PersistentClient(path="./chroma_db")
    collection = client.get_collection(name="crime_and_punishment")
    genai.configure(api_key=os.environ["GEMINI_API_KEY"])
    gen_model = genai.GenerativeModel(model_name="models/gemini-flash-lite-latest")
    supabase = create_client(os.environ["SUPABASE_URL"], os.environ["SUPABASE_KEY"])
    return embedder, collection, gen_model, supabase

embedder, collection, gen_model, supabase = load_resources()

def retrieve_context(query, top_k=5):
    query_embedding = embedder.encode([query]).tolist()
    results = collection.query(query_embeddings=query_embedding, n_results=top_k)
    return "\n\n---\n\n".join(results["documents"][0])

def generate_answer(query, chat_history=None):
    context = retrieve_context(query)
    history_text = ""
    if chat_history:
        for turn in chat_history:
            role = "User" if turn["role"] == "user" else "Assistant"
            history_text += f"{role}: {turn['content']}\n"

    prompt = f"""You are answering questions about the novel "Crime and Punishment" by Fyodor Dostoevsky.
Use ONLY the context below to answer. If the answer isn't in the context, say you don't know.

Context:
{context}

Conversation so far:
{history_text}
User: {query}
Assistant:"""
    response = gen_model.generate_content(prompt)
    return response.text

def load_history_from_supabase(session_id):
    resp = (
        supabase.table("chat")
        .select("role, message")
        .eq("session_id", session_id)
        .order("id")
        .execute()
    )
    return [
        {"role": "assistant" if row["role"] == "ai" else "user", "content": row["message"]}
        for row in resp.data
    ]

def save_message_to_supabase(session_id, role, message):
    supabase.table("chat").insert(
        {"session_id": session_id, "role": role, "message": message}
    ).execute()

mode = st.sidebar.radio(
    "Mode",
    ["Step 1: Single-turn", "Step 2: Multi-turn (in-memory)", "Step 3: Persistent (Supabase)"]
)
session_id = st.sidebar.text_input("Session ID (Step 3 only)", value="demo-session-1")

if "messages" not in st.session_state:
    st.session_state.messages = []

if mode.startswith("Step 3") and not st.session_state.get("_loaded"):
    st.session_state.messages = load_history_from_supabase(session_id)
    st.session_state._loaded = True

for msg in st.session_state.messages:
    with st.chat_message(msg["role"]):
        st.markdown(msg["content"])

user_input = st.chat_input("Ask something about Crime and Punishment...")

if user_input:
    with st.chat_message("user"):
        st.markdown(user_input)

    history = st.session_state.messages if not mode.startswith("Step 1") else None

    with st.spinner("Thinking..."):
        answer = generate_answer(user_input, chat_history=history)

    with st.chat_message("assistant"):
        st.markdown(answer)

    if not mode.startswith("Step 1"):
        st.session_state.messages.append({"role": "user", "content": user_input})
        st.session_state.messages.append({"role": "assistant", "content": answer})

    if mode.startswith("Step 3"):
        save_message_to_supabase(session_id, "user", user_input)
        save_message_to_supabase(session_id, "ai", answer)
```

# RAG Evaluation

## Step 4
Evaluate your RAG App using ragas:

- https://docs.ragas.io/en/stable/getstarted/evals/
- https://docs.ragas.io/en/stable/concepts/metrics/available_metrics/
- https://docs.ragas.io/en/stable/howtos/applications/

```python
EVAL_SET = [
    {
        "question": "Who is the main character of Crime and Punishment?",
        "ground_truth": "The main character is Rodion Raskolnikov, a poor former student in St. Petersburg."
    },
    {
        "question": "Why does Raskolnikov murder the pawnbroker?",
        "ground_truth": "Raskolnikov murders the old pawnbroker Alyona Ivanovna partly out of poverty and partly to test his theory that extraordinary people are morally entitled to transgress laws for a greater good."
    },
    {
        "question": "Who is Sonia and what role does she play in the novel?",
        "ground_truth": "Sonia Marmeladova is a young woman forced into prostitution to support her family. She becomes Raskolnikov's moral and spiritual guide, urging him toward confession and redemption."
    },
    {
        "question": "Who investigates the murder and suspects Raskolnikov?",
        "ground_truth": "Porfiry Petrovich, the police investigator, suspects Raskolnikov and psychologically pressures him toward confession."
    },
    {
        "question": "How does the novel end for Raskolnikov?",
        "ground_truth": "Raskolnikov confesses to the murder, is sentenced to penal servitude in Siberia, and begins a path toward spiritual redemption with Sonia's support."
    }
]

print(f"{len(EVAL_SET)} evaluation questions defined.")
```

### Collect answers + retrieved contexts

```python
import time

eval_rows = {"user_input": [], "response": [], "retrieved_contexts": [], "reference": []}

for item in EVAL_SET:
    print(f"Running: {item['question']}")
    context_chunks = retrieve_context_chunks(item["question"], top_k=5)

    for attempt in range(5):
        try:
            answer = generate_answer(item["question"])
            break
        except Exception as e:
            if "429" in str(e) or "quota" in str(e).lower():
                print("Rate limited, waiting 60s...")
                time.sleep(60)
            else:
                raise

    eval_rows["user_input"].append(item["question"])
    eval_rows["response"].append(answer)
    eval_rows["retrieved_contexts"].append(context_chunks)
    eval_rows["reference"].append(item["ground_truth"])
    time.sleep(15)

print("Done collecting answers.")
```

### Run RAGAS scoring


```python
from datasets import Dataset
from ragas import evaluate
from ragas.metrics import faithfulness, answer_relevancy, context_precision, context_recall
from ragas.llms import LangchainLLMWrapper
from ragas.embeddings import LangchainEmbeddingsWrapper
from ragas.run_config import RunConfig
from langchain_google_genai import ChatGoogleGenerativeAI, GoogleGenerativeAIEmbeddings
from langchain_core.rate_limiters import InMemoryRateLimiter
import os
from google.colab import userdata

os.environ["GOOGLE_API_KEY"] = userdata.get("GEMINI_API_Key")

rate_limiter = InMemoryRateLimiter(
    requests_per_second=0.15,   # ~1 call every ~6-7 seconds, safely under 15/min
    check_every_n_seconds=0.1,
    max_bucket_size=1,
)

eval_llm = LangchainLLMWrapper(
    ChatGoogleGenerativeAI(
        model="models/gemini-3.5-flash-lite",
        rate_limiter=rate_limiter,
    )
)
eval_embeddings = LangchainEmbeddingsWrapper(
    GoogleGenerativeAIEmbeddings(model="models/gemini-embedding-001")
)

run_config = RunConfig(max_workers=1, timeout=180, max_retries=10)

dataset = Dataset.from_dict(eval_rows)

results = evaluate(
    dataset,
    metrics=[faithfulness, answer_relevancy, context_precision, context_recall],
    llm=eval_llm,
    embeddings=eval_embeddings,
    run_config=run_config
)

print(results)
```
